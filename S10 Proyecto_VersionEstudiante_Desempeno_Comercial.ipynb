### Diseño y planificación
Planificación del dashboard

Para diseñar el dashboard decidí organizar la información en dos vistas complementarias, siguiendo una narrativa visual que facilite la interpretación de los datos. La Vista Overview, denominada Resumen Ejecutivo, presenta una visión general del desempeño del negocio mediante tres indicadores clave: Ingresos Totales, Rentabilidad Total y Unidades Vendidas, junto con un gráfico de líneas que muestra la evolución de los ingresos durante 2024 y 2025 y un gráfico de barras que permite comparar los ingresos totales entre Perú, Chile y Colombia. También incorporé un segmentador de año para facilitar la comparación entre ambos periodos.

En la Vista Detalle, denominada Análisis Detallado, organicé los elementos para profundizar en las diferencias observadas en los ingresos. Primero incorporé un gráfico de barras por segmento de clientes para identificar qué segmento genera mayores ingresos; después, un gráfico por categoría de productos para comparar su contribución; y finalmente, un gráfico de región y país para analizar las diferencias geográficas. Como complemento, incorporé una tabla de detalle de ventas con información de fecha_pedido, país, categoría_producto, segmento_cliente, unidades_vendidas e ingresos, permitiendo consultar las características de las ventas individuales. Además, incluí segmentadores de año y país para facilitar el análisis de los resultados.

La organización de ambas vistas busca mantener una lectura clara y evitar la sobrecarga de información. El Resumen Ejecutivo permite identificar rápidamente el comportamiento general de los ingresos, mientras que el Análisis Detallado permite explorar las principales diferencias según clientes, productos y ubicación geográfica.
